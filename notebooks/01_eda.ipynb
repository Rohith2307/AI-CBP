# Customer Sentiment Analysis — Exploratory Data Analysis (EDA)

This notebook provides systematic, data-driven exploration of the validated customer feedback dataset, examining:
1. Sentiment class distribution & class imbalance
2. Comment length distributions (character and word level)
3. Top characteristic keywords for Positive, Negative, and Neutral sentiment
4. Sentiment breakdown across products/brands
5. Temporal sentiment volume and negative feedback spike analysis
6. Root-cause complaint category breakdown

```python
import pandas as pd
from pathlib import Path
import src
from src.eda import (
    compute_sentiment_summary,
    compute_length_statistics,
    compute_top_terms,
    compute_product_insights,
    compute_temporal_insights,
    compute_complaint_topics,
    create_sentiment_bar_chart,
    create_sentiment_donut_chart,
    create_length_distribution_chart,
    create_top_words_bar_chart,
    create_product_sentiment_chart,
    create_temporal_trend_chart,
    create_complaint_topics_chart,
)

dataset_path = Path('../data/processed/validated_airline_sentiment.csv')
if not dataset_path.exists():
    dataset_path = Path('../data/sample/sample_airline_sentiment.csv')

df = pd.read_csv(dataset_path)
print(f"Loaded dataset with {len(df):,} records and {len(df.columns)} columns.")
```

## 1. Sentiment Class Distribution & Imbalance

```python
summary = compute_sentiment_summary(df)
print(summary['explanation'])
fig_bar = create_sentiment_bar_chart(df)
fig_bar.show()
```

## 2. Text Length Distributions

```python
length_stats = compute_length_statistics(df)
print(length_stats['explanation'])
fig_len = create_length_distribution_chart(df)
fig_len.show()
```

## 3. Characteristic Keywords by Sentiment

```python
for sentiment in ['Positive', 'Negative', 'Neutral']:
    fig_words = create_top_words_bar_chart(df, sentiment=sentiment, top_n=10)
    fig_words.show()
```

## 4. Product / Brand Sentiment Breakdown

```python
prod_insights = compute_product_insights(df)
if prod_insights:
    print(prod_insights['explanation'])
    fig_prod = create_product_sentiment_chart(df)
    fig_prod.show()
```

## 5. Temporal Volume Trends & Spikes

```python
temp_insights = compute_temporal_insights(df)
if temp_insights:
    print(temp_insights['explanation'])
    fig_temp = create_temporal_trend_chart(df)
    fig_temp.show()
```

## 6. Categorized Customer Complaint Topics

```python
topics = compute_complaint_topics(df)
if topics:
    print(topics['explanation'])
    fig_topics = create_complaint_topics_chart(df)
    fig_topics.show()
```
