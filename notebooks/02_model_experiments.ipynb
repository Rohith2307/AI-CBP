# Customer Sentiment Analysis — Model Training & Comparison Experiments

This notebook trains and compares three classical ML classifiers:
1. **Multinomial Naive Bayes** (Baseline text classification model)
2. **Logistic Regression** (Linear probabilistic model with balanced class weighting)
3. **Linear Support Vector Machine (LinearSVC)** (Maximum margin linear hyperplane classifier)

Features are extracted using word unigrams and bigrams via TF-IDF (fitted strictly on the training partition).
Models are compared across **Accuracy, Macro Precision, Macro Recall, Macro F1, 5-fold CV, and Inference Latency**.

```python
import pandas as pd
from pathlib import Path
import src
from src.train import train_and_compare_models

dataset_path = Path('../data/processed/preprocessed_airline_sentiment.csv')
if not dataset_path.exists():
    dataset_path = Path('../data/processed/validated_sample.csv')

df = pd.read_csv(dataset_path)
print(f"Loaded dataset with {len(df):,} records.")
```

## 1. Train and Compare Candidate Models

```python
experiment = train_and_compare_models(
    df=df,
    text_col='clean_text' if 'clean_text' in df.columns else 'text',
    sentiment_col='sentiment',
    output_dir='../models',
)

print("=== Model Comparison Summary ===")
display(experiment.comparison_table)
print(f"\nWinning Model: {experiment.best_model_name}")
print(f"Selection Rationale:\n{experiment.selection_rationale}")
```

## 2. Detailed Per-Class Breakdown

```python
for m in experiment.metrics_list:
    print(f"\n--- {m.model_name} Per-Class Metrics ---")
    cls_df = pd.DataFrame(m.class_metrics).T
    display(cls_df)
```
